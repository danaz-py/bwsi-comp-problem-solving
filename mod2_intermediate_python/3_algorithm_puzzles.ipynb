## Roman Numeral to Integer

### Q2. Write a function that converts a Roman numeral string into an integer.
Roman numerals convert very easily into integers: I is 1, V is 5, and X is 10. String them together like XVI, and the number is 16. However, when the numeral I is *in front* of another numeral, it will subract from the next numeral in the sequence. So IV is 4, while VI is 6. Your task is to write a function to account for this property.
```python
# Example
rom_to_int('IV') #returns 4
rom_to_int('VI') #returns 6
rom_to_int('XII') #returns 12
rom_to_int('XXIX') #returns 29
```

````python
# Write your funciton here!
def rom_to_int(string) -> int:
    input_roms = []
    number = 0
    index = 0
    for char_ in string:
        input_roms.append(char_)
    if len(input_roms) == 1:
        if input_roms[0] == "I":
            print("1")
        elif input_roms[0] == "V":
            print("5")
        elif input_roms[0] == "X":
            print("10")
    else:
        for char in input_roms:
            print("The char is" + char)
            #if there is I and the I after is not an I then the next numeral must be a V or X which would add 4 or 9 respectively
            if (char == "I") and input_roms[index + 1] != "I":
                if input_roms[index + 1] == "V":
                    number += 4
                elif input_roms[index + 1] == "X":
                    number += 9
            elif (char == "V") and input_roms[index + 1] == "I":
                number += 5
            elif (char == "X") and input_roms[index + 1] == "I":
                number += 10
            # #if it's only this at the end!
            elif (char == "I") and (all(y == "I" for y in list(input_roms[index-1:]))):
                number += (1 * len(list(input_roms[index:])))
                break
        if number > 7:
            print(number-1)
        else:
            print(number)
````
